```python
import pandas as pd

df = pd.read_csv('../data/listings.csv')
print(df.shape)
df.head()
```

```python
print(df.columns.tolist())
```

```python
key_columns = [
    'neighbourhood_cleansed', 'neighbourhood_group_cleansed', 'latitude', 'longitude',
    'property_type', 'room_type', 'accommodates', 'bedrooms', 'beds',
    'price', 'minimum_nights', 'availability_365',
    'number_of_reviews', 'review_scores_rating', 'host_is_superhost',
    'estimated_occupancy_l365d', 'estimated_revenue_l365d'
]

print(df[key_columns].isnull().sum())
print()
print(df[key_columns].dtypes)
```

```python
print(df['price'].head(10))
print(df['price'].dtype)
```

```python
print(df['neighbourhood_cleansed'].unique())
```

```python
print(df['neighbourhood_group_cleansed'].unique())
```

```python
# Clean price: strip $ and comma, convert to float
df['price_clean'] = df['price'].str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)

print(df['price_clean'].describe())
print("Nulls:", df['price_clean'].isnull().sum())
```

```python
# Check what's going on with null-price listings
null_price = df[df['price_clean'].isnull()]
print(null_price['has_availability'].value_counts())
print(null_price['availability_365'].describe())
```

```python
client_districts = ['Eixample', 'Gràcia', 'Ciutat Vella']
df['client_district'] = df['neighbourhood_group_cleansed'].isin(client_districts)

print(df['client_district'].value_counts())
```

```python
df['host_is_superhost'] = df['host_is_superhost'].map({'t': True, 'f': False})
print(df['host_is_superhost'].value_counts(dropna=False))
```

```python
# Drop rows with no price (not analyzable for pricing purposes)
df_clean = df[df['price_clean'].notnull()].copy()
print("Rows after dropping null prices:", df_clean.shape[0])

# Check room_type breakdown - we care about entire apartments (Marisol's unit type)
print(df_clean['room_type'].value_counts())
```

```python
df_clean['is_entire_apt'] = df_clean['room_type'] == 'Entire home/apt'
print(df_clean['is_entire_apt'].value_counts())

# Check for price outliers - that max of $10,542/night looks suspicious
print(df_clean['price_clean'].sort_values(ascending=False).head(20))
```

```python
print(df_clean['price_clean'].sort_values().head(20))
```

```python
high_price = df_clean.sort_values('price_clean', ascending=False).head(10)
print(high_price[['name', 'room_type', 'accommodates', 'bedrooms', 'price_clean', 'neighbourhood_group_cleansed']])
```

```python
low_price = df_clean.sort_values('price_clean').head(10)
print(low_price[['name', 'room_type', 'accommodates', 'bedrooms', 'price_clean', 'neighbourhood_group_cleansed']])
```

```python
# Remove the one clear price data error (1BR/2-guest loft at $10,542/night)
df_clean = df_clean[df_clean['price_clean'] < 10000].copy()
print("Rows after removing extreme outlier:", df_clean.shape[0])
```

```python
df_clean.to_csv('../data/listings_cleaned.csv', index=False)
print("Saved listings_cleaned.csv")
print(df_clean.shape)
```

```python
relevant_cols = [
    'id', 'name', 'neighbourhood_cleansed', 'neighbourhood_group_cleansed',
    'latitude', 'longitude', 'property_type', 'room_type', 'is_entire_apt',
    'client_district', 'accommodates', 'bedrooms', 'beds', 'price_clean',
    'minimum_nights', 'availability_365', 'number_of_reviews',
    'review_scores_rating', 'host_is_superhost', 'estimated_occupancy_l365d',
    'estimated_revenue_l365d'
]

df_sql = df_clean[relevant_cols].copy()
df_sql.to_csv('../data/listings_for_sql.csv', index=False)
print(df_sql.shape)
df_sql.head()
```
